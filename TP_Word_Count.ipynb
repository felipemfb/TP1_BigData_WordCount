L'objectif de ce travail est de mettre en œuvre quatre approches visant à compter le nombre d'occurrences de chaque mot dans un fichier .txt, tout en comprenant le fonctionnement de chacune d'elles.

Les 4 versions proposées sont :

- Version simple utilisant Python

- Pseudo-MapReduce

- Hadoop MapReduce

- **Approche utilisant PySpark**

Pour évaluer chaque approche, le temps d'exécution de chacune sera calculé en utilisant des fichiers texte contenant des nombres de mots différents.

## Importation des bibliothèques utilisées

```python
from pathlib import Path
import random
from time import perf_counter
import pandas as pd
from collections import defaultdict
from pyspark.sql import SparkSession
from pyspark.sql.functions import split
from pyspark import SparkContext
```

## Création des fichiers txt

Quatre fichiers .txt contenant des mots aléatoires sur chaque ligne ont été créés. Chaque fichier contient dix fois plus de mots que le précédent, dans le but d'évaluer les performances de chaque approche pour différentes tailles de fichiers d'entrée.

```python
n = 1000
tailles = [1, 10, 100, 1000]
mots = ["chat", "chien", "arbre", "fleur", "maison", "soleil", "montagne", "livre", "ciel", "mer", "lune", "route", "oiseau", "jardin", "voiture"]

dossier = Path("data")
dossier.mkdir(exist_ok=True)

fichiers = {}

for m in tailles:
    lignes = n * m
    chemin = dossier / f"mots_taille_{m}.txt"

    with chemin.open("w", encoding="utf-8") as fichier:
        for i in range(lignes):
            mot = random.choice(mots)
            fichier.write(mot + "\n")

    fichiers[m] = chemin

    print(f"{chemin.name} : {lignes} lignes, {chemin.stat().st_size} bytes")
```

## Version simple utilisant Python

```python
def compter_occurrences(fichier):
    with open(fichier, 'r', encoding = 'utf-8') as f:
        mots = f.read().split()

    compteur = {}
    for mot in mots:
        compteur [mot] = compteur.get(mot, 0) + 1
    return compteur

if __name__ == "__main__":
    dossier = Path("data")
    fichiers = sorted(dossier.glob("*.txt"))
    for fichier in fichiers:
        start = perf_counter()
        resultats = compter_occurrences(fichier)
        end = perf_counter()
        temps = end - start
        print(f"fichier : {fichier.name}\ntemps : {temps:.6f} secondes\nnombre de mots : {sum(resultats.values())}\nnombre de mots uniques : {len(resultats)}\n\nOccurrences :")

        for mot, nb in sorted(resultats.items(), key = lambda x : x [1], reverse = True):
            print (f"{mot} : {nb}")
        print("\n\n\n")
```

## Pseudo-MapReduce

```python
def mapper(ligne):
    mots = ligne.strip().split()
    return [(mot, 1) for mot in mots]

def shuffle(mapped_data):
    grouped = defaultdict(list)
    for mot, valeur in mapped_data:
        grouped[mot].append(valeur)
    return grouped

def reducer(grouped_data):
    reduced = {}
    for mot, valeurs in grouped_data.items():
        reduced [mot] = sum(valeurs)
    return reduced

if __name__ == "__main__":
    dossier = Path("data")
    fichiers = sorted(dossier.glob("*.txt"))
    for fichier in fichiers:
        mapped = []
        start = perf_counter()
        with open(fichier, "r", encoding = "utf-8") as f:
            for ligne in f:
                mapped.extend(mapper(ligne))

        grouped = shuffle(mapped)
        reduced = reducer(grouped)

        end = perf_counter()
        temps = end - start
        print(f"fichier : {fichier.name}\ntemps : {temps:.6f} secondes\nnombre de mots : {sum(reduced.values())}\nnombre de mots uniques : {len(reduced)}\n\nOccurrences :")

        for mot, count in sorted(reduced.items(), key = lambda x : x[1] , reverse = True):
            print (f"{mot} : {count}")
        print("\n\n\n")
```

## Approche utilisant PySpark

```python
spark = SparkSession.builder.appName("CompteurDeMots").master("local[*]").config("spark.hadoop.fs.defaultFS", "file:///").getOrCreate()

sc = spark.sparkContext
dossier = Path("data")
fichiers = sorted(dossier.glob("*.txt"))

for fichier in fichiers:
    start = perf_counter()
    text_file = sc.textFile(str(fichier))

    counts = text_file.flatMap(lambda line: line.split()).map(lambda word: (word, 1)).reduceByKey(lambda a, b: a + b)

    counted = counts.collect()

    end = perf_counter()
    temps = end - start
    print(f"fichier : {fichier.name}\ntemps : {temps:.6f} secondes\nnombre de mots : {sum(count for mot, count in counted)}\nnombre de mots uniques : {len(counted)}\n\nOccurrences :")

    for mot, count in sorted(counted, key = lambda x : x[1] , reverse = True):
        print (f"{mot} : {count}")
    print("\n\n\n")
```

## Analyse des performances et conclusion

L'approche simple utilisant Python est la méthode la plus rapide parmi celles présentées pour les données utilisées. Le pseudo-MapReduce présente un temps d'exécution légèrement plus long en raison de l'utilisation des étapes de Map, Shuffle et Reduce, suivi par l'approche utilisant PySpark, qui introduit les côuts et le temps nécessaires à la préparation et à l’exécution des opérations Spark. Hadoop MapReduce est considéré comme la solution la plus stable pour des volumes de données extrêmement importants, mais ils impliquent un temps d'exécution de base élevé dû à la complexité de l'infrastructure.

Ce TP a permis de comprendre quatre approches de complexités différentes. Le choix de la méthode dépend directement de la taille de l'ensemble de données utilisé. Les petits jeux de données conviennent mieux au pseudo-MapReduce et, surtout, à l'approche simple utilisant Python, tandis que le traitement de Big Data justifie le recours à Hadoop MapReduce. PyShark agit comme un intermédiaire, car il dispose d'une meilleure infrastructure que le pseudo-MapReduce et que l'approche basée sur Python pour la gestion du Big Data, tout en offrant un temps d'exécution inférieur à celui de Hadoop MapReduce.
